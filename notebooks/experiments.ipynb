## Test 5: Agent Workflow (selective)

```python
# Test with selective agents (faster)
orchestrator = create_orchestrator()

query = "What is the derivative of sin(x)?"
print(f"Query: {query}")
print("\nRunning research with [planner, mathematician]...\n")

context = orchestrator.run_research_selective(
    query,
    agents=["planner", "mathematician"]
)

print(f"\nExecution path: {' -> '.join(context.execution_path)}")
print(f"Mathematical insights: {len(context.mathematical_insights)}")
print(f"Equations: {len(context.final_equations)}")
print(f"Errors: {len(context.errors)}")

if context.mathematical_insights:
    print(f"\nFirst insight:\n{context.mathematical_insights[0][:200]}...")
```
